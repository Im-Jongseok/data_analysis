로컬 실행 결과를 저장했으며, 마지막 셀은 `iris_model.pkl`을 random forest로 갱신합니다.

```python
import sklearn
print(sklearn.__version__)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

iris = load_iris(as_frame=True)
X = iris.frame[iris.feature_names]
y = iris.target_names[iris.target]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0)

model = RandomForestClassifier(n_estimators=100, random_state=0)
model.fit(X_train, y_train)
print("training :", model.score(X_train, y_train), " test :", model.score(X_test, y_test))
```

```python
new = pd.DataFrame([[6.0, 2.9, 4.5, 1.5]], columns=iris.feature_names)
print(model.predict(new)[0], model.predict_proba(new).round(2))
```

```python
import joblib

joblib.dump(model, "iris_model.pkl")
try:
    from google.colab import files
    files.download("iris_model.pkl")
except ImportError:
    pass
```
